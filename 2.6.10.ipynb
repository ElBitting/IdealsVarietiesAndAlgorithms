```sage
R.<x, y, z> = PolynomialRing(QQ, order= "deglex")
```

```sage
f_1 = x^2 - y
f_2 = x^3 - z
G = [f_1,f_2]
```

```sage
def mult_deg(f): 
    return list(f.lt().degrees())
```

```sage
def lcm(f,g):
    gamma = [0,0,0]
    i = 0
    while i < 3:
        if mult_deg(f)[i] > mult_deg(g)[i]:
            gamma[i] += mult_deg(f)[i]
        else:
            gamma[i] += mult_deg(g)[i]
        i+=1
    return(x^gamma[0]*y^gamma[1]*z^gamma[2])
```

```sage
def S_polynomial(f,g):
    return(lcm(f,g)/f.lt()*f-lcm(f,g)/g.lt()*g)
```

```sage
def div_alg(p, G):
    r = p
    quo = []
    for q in G:
        if r != 0:
            Q, R = r.quo_rem(q)
            quo.append(Q)
            r = R
    return (quo, r)
```

```sage
def is_Grobner(G):
    i = 0
    j = 1
    S = []
    while i < len(G):
        while j < len(G):
            S.append(S_polynomial(G[i],G[j]))
            j+=1
        i+=1
        j=i+1
    for s in S:
        q, r = div_alg(s,G)
        if r != 0:
            return (false, S.index(s), s, div_alg(s, G)[1])
    return true
            
```

```sage
is_Grobner(G)
```
